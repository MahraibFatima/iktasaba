# with Runnable
abstract method - invoke

```python
from abc import ABC, abstractmethod
class Runnable(ABC):
  @abstractmethod
  def invoke(input_data):
    pass
```

```python
class DummyLLM_withRunnable(Runnable):
  def __init__(self):
    print("LLM initialized.")
  def invoke(self, prompt):
    responseList= [
         "Ali",
    'Abdullah',
    "Hashim",
    "Akbar"
]
    return {'response':random.choice(responseList)}
  def predict(self, prompt):
    responseList= [
         "Ali",
    'Abdullah',
    "Hashim",
    "Akbar"
]
    return {'response':random.choice(responseList)}
```

```python
class DummyPromptTemplate_withRunnable(Runnable):
  def __init__(self, template, input_variables):
    self.template= template
    self.input_variables= input_variables
  def invoke(self, input_dict):
    return self.template.format(**input_dict)
  def format(self, input_dict):
    return self.template.format(**input_dict)
```

```python
class DummayStrOutputParser_withRunnable(Runnable):
  def __init__(self) -> None:
    pass
  def invoke(self, input_data):
    return input_data['response']
```

```python
class DummyRunnableConnector(Runnable):
  def __init__(self, runnable_list):
    self.runnable_list= runnable_list
  def invoke(self, input_data):
    for runnable in self.runnable_list:
      input_data= runnable.invoke(input_data)
    return input_data
```

```python
template= DummyPromptTemplate_withRunnable(
    template="write {length} peom about the {topic}.",
    input_variables=['topic', 'length']
)
```

```python
llm= DummyLLM_withRunnable()
```

```python
parser= DummayStrOutputParser_withRunnable()
```

```python
chain= DummyRunnableConnector([template, llm, parser])
```

```python
chain.invoke({'length':'long', 'topic':'forest'})
```

```python
template1 = DummyPromptTemplate_withRunnable(
    template='Write a joke about {topic}',
    input_variables=['topic']
)
```
